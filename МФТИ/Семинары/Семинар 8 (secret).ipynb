```python
import json

import numpy as np
```

```python
eps = np.random.normal(size=5000)
```

```python
y1 = np.array([eps[0]])
y2 = np.array([-2 + eps[0]])
y3 = np.array([10 + eps[0]])
y15 = -10 + eps[0:1]
y21 = -10 + eps[0:1]

for t in range(1, 5000):
    y1 = np.append(y1, 0.99 * y1[-1] + eps[t])
    y2 = np.append(y2, -2 - 0.99 * y2[-1] + eps[t])
    y3 = np.append(y3, 10 + 0.8 * y3[-1] + eps[t])
    y15 = np.append(y15, -10 + eps[t] + eps[t - 1])
    y21 = np.append(y21, -10 + 0.8 * y21[-1] + eps[t] + 0.5 * eps[t - 1])
```

```python
y8 = eps[0:2]
y10 = -5 + eps[0:2]
y12 = 10 + eps[0:2]
y19 = -2 + eps[0:2]

for t in range(2, 5000):
    y8 = np.append(y8, 0.4 * y8[-1] + 0.4 * y8[-2] + eps[t])
    y10 = np.append(y10, -5 + y10[-1] - 0.25 * y10[-2] + eps[t])
    y12 = np.append(y12, 10 + 1.5 * y12[-1] - 0.9 * y12[-2] + eps[t])
    y19 = np.append(y19, -2 + eps[t] - eps[t - 2])
```

```python
y28 = -10 + eps[0:4]

for t in range(4, 5000):
    y28 = np.append(y28, -10 + 0.8 * y28[-4] + eps[t])
```

```python
data = {}
data[1] = y1.tolist()[500:]
data[2] = y2.tolist()[500:]
data[3] = y3.tolist()[500:]
data[8] = y8.tolist()[500:]
data[10] = y10.tolist()[500:]
data[12] = y12.tolist()[500:]
data[15] = y15.tolist()[500:]
data[19] = y19.tolist()[500:]
data[21] = y21.tolist()[500:]
data[28] = y28.tolist()[500:]
```

```python
with open("Семинар 6.json", "w") as f:
    json.dump(data, f)
```
