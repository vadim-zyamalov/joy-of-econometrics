```python
import json
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pmdarima as pmd
import statsmodels.graphics.tsaplots as tsp
```

```python
def acf_pacf(y):
    plt.clf()
    _, axes = plt.subplots(1, 2)
    tsp.plot_acf(y, zero=False, bartlett_confint=False, ax=axes[0])
    tsp.plot_pacf(y, zero=False, ax=axes[1], method="ywm")
    plt.show()
```

```python
with open("Семинар 5.json", "r") as f:
    data = json.load(f)

npdata = {}
for k, v in data.items():
    npdata[int(k)] = np.array(v)
```

```python
print(npdata.keys())
```

```python
plt.plot(npdata[8])
```

```python
acf_pacf(npdata[28])
```

```python
model = pmd.auto_arima(npdata[21])
model.order
```
